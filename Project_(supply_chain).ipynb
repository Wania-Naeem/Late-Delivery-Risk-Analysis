```python
# Define the columns to drop due to excessive missing values or privacy concerns
columns_to_drop_further = [
    'Product Description', # 100% missing
    'Order Zipcode',       # Mostly missing
    'Customer Password',   # Sensitive personal identifier
    'Customer Email',      # Sensitive personal identifier
    'Customer Fname',      # Sensitive personal identifier
    'Customer Lname',      # Sensitive personal identifier
    'Customer Street'      # Sensitive personal identifier
]

# Drop the specified columns from the DataFrame
df.drop(columns=columns_to_drop_further, inplace=True)

# Convert date columns to datetime format
df['order date (DateOrders)'] = pd.to_datetime(df['order date (DateOrders)'])
df['shipping date (DateOrders)'] = pd.to_datetime(df['shipping date (DateOrders)'])

# Check for any remaining missing values and drop those rows if they exist
# This handles any `NaN` values that might appear after date conversion or other issues
if df.isnull().sum().sum() > 0:
    initial_rows = df.shape[0]
    df.dropna(inplace=True)
    final_rows = df.shape[0]
    print(f"Dropped {initial_rows - final_rows} rows due to remaining missing values.")

# Print the final shape of the cleaned dataframe
print(f"Final DataFrame shape after cleaning: {df.shape}")

# Print the list of remaining column names
print("Remaining column names after cleaning:")
print(df.columns.tolist())
```

```python
# Drop remaining unnecessary columns: duplicates, pure identifiers, constant values, 
# and high-cardinality text columns not used in modeling
columns_to_drop_final = [
    'Product Status',           # Constant value (always 0) — no predictive value
    'Order Customer Id',        # Duplicate of 'Customer Id'
    'Order Item Cardprod Id',   # Duplicate of 'Product Card Id'
    'Product Category Id',      # Duplicate of 'Category Id'
    'Order Item Id',            # Pure row identifier — no pattern to learn
    'Order Id',                 # Pure row identifier
    'Department Id',            # Redundant with 'Department Name'
    'Product Name',             # High-cardinality text, not used in features
    'Customer City',            # High-cardinality, not used in features
    'Order City'                # High-cardinality, not used in features
]

df.drop(columns=columns_to_drop_final, inplace=True)
print(f"Shape after dropping final unnecessary columns: {df.shape}")
print("Remaining columns:", df.columns.tolist())
```

### Feature Engineering: Adding New Columns

```python
# 1. Create three new columns from 'order date (DateOrders)'
df['order_month'] = df['order date (DateOrders)'].dt.month
df['order_day_of_week'] = df['order date (DateOrders)'].dt.dayofweek
df['order_quarter'] = df['order date (DateOrders)'].dt.quarter

print("New date features created: 'order_month', 'order_day_of_week', 'order_quarter'")
```

### Redefining Features and Target

```python
additional_features = [
    'Type',
    'Order Country',
    'Order State',
    'Product Price',
    'Benefit per order'
]

original_features = [
    'Days for shipment (scheduled)', 'Shipping Mode', 'Order Region', 'Market',
    'Category Name', 'Customer Segment', 'Order Item Quantity', 'Sales',
    'Order Item Discount Rate'
]
new_date_features = ['order_month', 'order_day_of_week', 'order_quarter']

full_feature_list = original_features + new_date_features + additional_features
target = 'Late_delivery_risk'

X = df[full_feature_list].copy()
y = df[target]

print(f"New X (features) DataFrame shape: {X.shape}")
print("New list of feature column names:")
print(X.columns.tolist())
```

### One-Hot Encoding and Data Splitting

```python
from sklearn.model_selection import train_test_split

categorical_cols_for_ohe = [
    'Shipping Mode', 'Order Region', 'Market', 'Category Name', 
    'Customer Segment', 'Type', 'Order Country', 'Order State'
]
categorical_cols_present = [col for col in categorical_cols_for_ohe if col in X.columns]

X_encoded_categorical = pd.get_dummies(X[categorical_cols_present], drop_first=True)
numeric_cols = [col for col in X.columns if col not in categorical_cols_present]
X_numeric = X[numeric_cols]

X_encoded = pd.concat([X_numeric, X_encoded_categorical], axis=1)

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.2, random_state=42, stratify=y
)

print(f"Final X_encoded shape: {X_encoded.shape}")
```

### Model Training and Prediction

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

rf_model = RandomForestClassifier(
    n_estimators=100,
    max_depth=15,          # Trees ko reasonable size tak limit kar diya
    min_samples_leaf=5,    # Extra safety — overfitting kam karega
    random_state=42,
    n_jobs=-1
)
rf_model.fit(X_train, y_train)

rf_predictions = rf_model.predict(X_test)
rf_accuracy = accuracy_score(y_test, rf_predictions)
print(f"New RandomForest Accuracy (max_depth=15): {rf_accuracy:.4f}")
```

```python
# 3. Train an XGBClassifier
print("\nTraining XGBClassifier...")
# eval_metric='logloss' is specified to avoid a warning in newer XGBoost versions
# use_label_encoder=False is used to suppress a common warning for binary classification
xgb_model = XGBClassifier(random_state=42, eval_metric='logloss', use_label_encoder=False)
xgb_model.fit(X_train, y_train)
print("XGBClassifier trained.")

# 4. Generates predictions on X_test
xgb_predictions = xgb_model.predict(X_test)

# 5. Generates predicted probabilities
xgb_probabilities = xgb_model.predict_proba(X_test)
print("XGBClassifier predictions and probabilities generated.")
```

```python
# 6. Prints a confirmation message
print("\nBoth models are trained and predictions/probabilities are made.")
```

### Model Evaluation

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# --- Evaluation for RandomForestClassifier ---
print("\n--- RandomForestClassifier Evaluation ---")

# 1. Calculate and print accuracy_score
rf_accuracy = accuracy_score(y_test, rf_predictions)
print(f"RandomForestClassifier Accuracy: {rf_accuracy:.4f}")

# 2. Generate and visualize confusion matrix
rf_cm = confusion_matrix(y_test, rf_predictions)
plt.figure(figsize=(6, 5))
sns.heatmap(rf_cm, annot=True, fmt='d', cmap='Blues')
plt.title('RandomForestClassifier Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()

# 3. Print full classification_report
print("RandomForestClassifier Classification Report:")
print(classification_report(y_test, rf_predictions))

# 4. Extract and plot ONLY TOP 20 feature_importances_ (fixed — readable now)
print("\nRandomForestClassifier Feature Importances (Top 20):")
feature_importances = pd.Series(rf_model.feature_importances_, index=X_train.columns)
top_20_features = feature_importances.sort_values(ascending=False).head(20)

plt.figure(figsize=(10, 8))
top_20_features.sort_values().plot(kind='barh', color='steelblue')
plt.title('Top 20 Most Important Features (RandomForestClassifier)')
plt.xlabel('Importance')
plt.ylabel('Feature')
plt.tight_layout()
plt.show()
```

```python
# --- Evaluation for XGBClassifier ---
print("\n--- XGBClassifier Evaluation ---")

# 1. Calculate and print accuracy_score
xgb_accuracy = accuracy_score(y_test, xgb_predictions)
print(f"XGBClassifier Accuracy: {xgb_accuracy:.4f}")

# 2. Generate and visualize confusion matrix
xgb_cm = confusion_matrix(y_test, xgb_predictions)
plt.figure(figsize=(6, 5))
sns.heatmap(xgb_cm, annot=True, fmt='d', cmap='Greens')
plt.title('XGBClassifier Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()

# 3. Print full classification_report
print("XGBClassifier Classification Report:")
print(classification_report(y_test, xgb_predictions))
```

```python
# 5. Create a simple pandas DataFrame comparing both models' accuracy scores
model_comparison = pd.DataFrame({
    'Model': ['RandomForestClassifier', 'XGBClassifier'],
    'Accuracy': [rf_accuracy, xgb_accuracy]
})

print("\n--- Model Accuracy Comparison ---")
display(model_comparison.sort_values(by='Accuracy', ascending=False))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Create a pandas DataFrame comparing all four models' names and their accuracy scores
model_names = [
    'RandomForestClassifier',
    'XGBClassifier (Default)',
    'XGBClassifier (Tuned)',
    'LGBMClassifier'
]

accuracy_scores = [
    rf_accuracy_fresh, # From previous RandomForest evaluation
    xgb_accuracy_fresh, # From previous default XGBoost evaluation
    new_xgb_accuracy, # From previous tuned XGBoost evaluation
    lgbm_accuracy # From previous LGBMClassifier evaluation
]

all_models_comparison = pd.DataFrame({
    'Model': model_names,
    'Accuracy': accuracy_scores
})

# Sort from highest to lowest accuracy
all_models_comparison = all_models_comparison.sort_values(by='Accuracy', ascending=False).reset_index(drop=True)

# 2. Displays it as a clean table
print("\n--- All Models Accuracy Comparison ---")
display(all_models_comparison)

# 3. Creates a horizontal bar chart using matplotlib comparing all four models' accuracy
plt.figure(figsize=(10, 6))
sns.barplot(x='Accuracy', y='Model', data=all_models_comparison, palette='viridis')

# Highlight the best-performing model's bar in a different color
best_model_accuracy = all_models_comparison.iloc[0]['Accuracy']
best_model_name = all_models_comparison.iloc[0]['Model']

# Find the bar corresponding to the best model and set its color
for i, (model, accuracy) in enumerate(zip(all_models_comparison['Model'], all_models_comparison['Accuracy'])):
    if model == best_model_name:
        plt.barh(i, accuracy, color='red') # Highlight best model in red
    else:
        plt.barh(i, accuracy, color=sns.color_palette('viridis')[i]) # Use original palette for others

# Add accuracy values labeled directly on each bar
for index, row in all_models_comparison.iterrows():
    plt.text(row['Accuracy'], index, f'{row['Accuracy']:.4f}', va='center', ha='left')

plt.title('Model Accuracy Comparison')
plt.xlabel('Accuracy')
plt.ylabel('Model')
plt.xlim(0, 1) # Set x-axis limit from 0 to 1 for accuracy
plt.tight_layout()
plt.show()

# 4. Prints a summary statement identifying the best model by name and its accuracy
print(f"\nThe best-performing model is the {best_model_name} with an accuracy of {best_model_accuracy:.4f}.")
```
